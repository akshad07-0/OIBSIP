   Objective: Train a machine learning model to predict the species of an iris flower from its measurements.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))
df.head()
```

```python
# Dataset ki shape (rows, columns)
print("Shape:", df.shape)
```

```python
# Har column ka data type aur non-null count
df.info()
```

```python
# Har column mein missing values
df.isnull().sum()
```

```python
# Mean, std, min, max, quartiles
df.describe()
```

```python
# Har species ke kitne flowers hain
df['species'].value_counts()
```
